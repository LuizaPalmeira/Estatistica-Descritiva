```python
import pandas as pd # biblioteca para manipulação de dados 
import numpy as np # biblioteca para operações numéricas

# Definindo número aleatório para reprodutibilidade dos dados
np.random.seed(42)

# Definindo o volume de linhas do arquivo
TOTAL_LINHAS = 1000

# Lista de categorias e produtos para sorteio
categorias = ['Eletrônicos', 'Móveis', 'Informática', 'Eletrodomésticos']
cidades = ['Rio de Janeiro', 'Niterói', 'Petrópolis', 'Volta Redonda', 'Macaé']

# Gerando dados sintéticos
dados_sinteticos = {
    'ID_Transacao': [f"TXN-{10000 + i}" for i in range(TOTAL_LINHAS)],
    'Data_Venda': pd.date_range(start='2024-01-01', periods=TOTAL_LINHAS, freq='4h'),
    'Cidade': np.random.choice(cidades, size=TOTAL_LINHAS, p=[0.4, 0.2, 0.15, 0.15, 0.1]),
    'Categoria': np.random.choice(categorias, size=TOTAL_LINHAS),
    'Quantidade': np.random.randint(1, 10, size=TOTAL_LINHAS),
    'Preco_Unitario': np.round(np.random.uniform(50.0, 3500.0, size=TOTAL_LINHAS), 2)
}

# Criando o DataFrame
df_gerado = pd.DataFrame(dados_sinteticos)

# Calculando a coluna do Valor Total
df_gerado['Valor_Total'] = df_gerado['Quantidade'] * df_gerado['Preco_Unitario']

```

Exercício 1: Análise de Vendas
Enunciado:
Crie um programa que analise um conjunto de dados de vendas. O programa deve ser estruturado em funções e seguir as instruções abaixo:

Função le_dados_vendas(nome_arquivo): Esta função deve ler um arquivo CSV contendo informações de vendas com as colunas: Produto, Quantidade, Preço e retornar um DataFrame.

Função calcula_receita(df): Esta função deve receber um DataFrame e calcular a receita total de cada produto (multiplicando Quantidade por Preço). A receita deve ser adicionada como uma nova coluna chamada Receita.

Função total_vendas_por_produto(df): Esta função deve agrupar os dados por produto e calcular o total de vendas (soma das quantidades) e a receita total (soma das receitas). Retorne um novo DataFrame contendo o total de vendas e a receita por produto.

Função filtra_produtos_altas_vendas(df, limite): Esta função deve filtrar o DataFrame para retornar apenas os produtos cuja receita total é superior a um valor limite passado como argumento.

Função analise_vendas(nome_arquivo): Esta função deve orquestrar as chamadas das funções anteriores, retornando o DataFrame final com os produtos que geraram mais de R$ 500,00 em receita.

```python
import pandas as pd


# Lendo um arquivo de vendas e transformando em DataFrame
def le_dados_vendas(nome_arquivo):
    df_venda = pd.read_csv(f'{nome_arquivo}', sep=',') # Le e ja transforma em DataFrame, nao precisa do 'pd.DataFrame'
    return df_venda


# Calculo da nova coluna Receita (Q * P)
def calcula_receita(df):
    df['Receita'] = df['Quantidade'] * df['Preco']
    return df


# Agrupo dados dos produtos, calculo valor total das vendas e receitas, crio um novo df com elas
def total_vendas_por_produto(df):
   vendas_total = df.groupby('Produto')['Quantidade'].sum().reset_index() # Grupo pelo produto, pego quantidade e somo
   receitas_total = df.groupby('Produto')['Receita'].sum().reset_index() # Grupo pelo produto, pego receita e somo

    # Transformo os dois em df
   df_vendas_total = pd.DataFrame(vendas_total) 
   df_receitas_total = pd.DataFrame(receitas_total)

    # Junto eles atraves do produto (intersecção) em um unico df
   df_fusao = pd.merge(
       df_vendas_total,
       df_receitas_total,
       on='Produto', # Intersecção, o que eles tem em comum
   )
   return df_fusao #chamo o cara


# Filtro produtos cujo a receita total (2 df) e maior que 'limite'
def filtra_produtos_altas_vendas(df, limite):
    filtro = df[df['Receita'] > limite]
    return filtro

# Chamo todas as funcoes
def analise_vendas():
    df_venda = le_dados_vendas('c:\\Users\\palme\\Documents\\python\\UC2\\DATA\\vendas.csv')
    df_venda = calcula_receita(df_venda)
    df_fusao = total_vendas_por_produto(df_venda)
    df_filtrado = filtra_produtos_altas_vendas(df_fusao, 500)
    return df_filtrado

# Exibindo receitas com total > 500
resultado = analise_vendas()
display(resultado)
```

Exercício 2: Análise de Notas de Alunos
Enunciado:
Desenvolva um programa que analise as notas de alunos. O programa deve ser estruturado em funções e seguir as instruções abaixo:

Função le_dados_alunos(nome_arquivo): Esta função deve ler um arquivo CSV contendo informações de alunos com as colunas: Nome, Nota1, Nota2, Nota3 e retornar um DataFrame.

Função calcula_media(df): Esta função deve calcular a média das notas de cada aluno e adicionar uma nova coluna chamada Media ao DataFrame.

Função verifica_aprovacao(df): Esta função deve adicionar uma nova coluna chamada Aprovado, que deve conter True se a média do aluno for maior ou igual a 7 e False caso contrário.

Função separa_alunos_aprovados_reprovados(df): Esta função deve separar os alunos em dois DataFrames: um para aprovados e outro para reprovados, com base na coluna Aprovado.

Função analise_notas(nome_arquivo): Esta função deve orquestrar as chamadas das funções anteriores e retornar os DataFrames de alunos aprovados e reprovados.

```python
import pandas as pd

# Lendo um arquivo de notas
def le_dados_alunos(nome_arquivo):
    df_alunos = pd.read_csv(f'{nome_arquivo}', sep=',')
    return df_alunos


# Calcula a media das notas e cria uma nova coluna chamada Media
def calcula_media(df):
    df['Media'] = ((df['Nota1'] + df['Nota2'] + df['Nota3'])/ 3).round(1) # Round arredondo em uma casa decimal
    return df


# Se media for maior ou igual a 7 aparece True, se nao False
def verifica_aprovacao(df):
    df['Aprovado'] = df['Media'] >= 7 # Boleana automatico
    return df


# Separo alunos em dois df (aprovados e reprovados) com base na coluna aprovados
def separa_alunos_aprovados(df):
    aprovados = df[df['Aprovado'] == True]
    return aprovados

def separa_alunos_reprovados(df):
    reprovados = df[df['Aprovado'] == False]
    return reprovados


# Todas funcoes e mostro o df de alunos aprovados vs reprovados
def analise_notas():
    df_alunos = le_dados_alunos('c:\\Users\\palme\\Documents\\python\\UC2\\DATA\\notas.csv')
    df_alunos = calcula_media(df_alunos)
    df_alunos = verifica_aprovacao(df_alunos)
    aprovados = separa_alunos_aprovados(df_alunos)
    reprovados = separa_alunos_reprovados(df_alunos)
    return aprovados, reprovados

# Exibindo alunos aprovados e reprovados
aprovados, reprovados = analise_notas()
display(aprovados)
display(reprovados)
```

Exercício 3: Análise de Dados de Funcionários
Enunciado:
Crie um programa que analise os dados de funcionários de uma empresa. O programa deve ser estruturado em funções e seguir as instruções abaixo:

Função le_dados_funcionarios(nome_arquivo): Esta função deve ler um arquivo CSV com as colunas: Nome, Departamento, Salário e retornar um DataFrame.

Função calcula_salario_medio(df): Esta função deve agrupar os dados por departamento e calcular o salário médio de cada um, retornando um novo DataFrame com o nome do departamento e o salário médio.

Função filtra_departamentos_alto_salario(df, limite): Esta função deve filtrar o DataFrame para retornar apenas os departamentos cuja média salarial é superior a um valor limite passado como argumento.

Função analise_funcionarios(nome_arquivo): Esta função deve orquestrar as chamadas das funções anteriores e retornar o DataFrame final com os departamentos que têm um salário médio acima de R$ 5000,00.

```python

import pandas as pd

# 1. Lendo o arquivo de funcionários
def le_dados_funcionarios(nome_arquivo):
    df_funcionarios = pd.read_csv(nome_arquivo, sep=',', encoding='utf-8')
    return df_funcionarios


# 2. Calculando o salário médio por departamento
def calcula_salario_medio(df):
    df_media = df.groupby('Departamento')['Salário'].mean().reset_index()
    df_media = df_media.rename(columns={'Salário': 'Salário_Médio'})
    return df_media


# 3. Filtrando departamentos com salário médio acima do limite
def filtra_departamentos_alto_salario(df, limite):
    filtro = df[df['Salário_Médio'] > limite]
    return filtro


# 4. Organizando a análise completa
def analise_funcionarios():
    df_funcionarios = le_dados_funcionarios('c:\\Users\\palme\\Documents\\python\\UC2\\DATA\\funcionarios.csv')
    df_media = calcula_salario_medio(df_funcionarios)
    resultado = filtra_departamentos_alto_salario(df_media, 0)
    return resultado


analise_resultado = analise_funcionarios()
display(analise_resultado)
```
